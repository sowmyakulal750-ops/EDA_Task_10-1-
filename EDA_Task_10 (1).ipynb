```python
plt.figure(figsize=(10,6))

sns.regplot(
    data=df,
    x="sqft_living",
    y="price"
)
plt.title("scatter plots with regression trend line")
plt.xlabel("sqft_living")
plt.ylabel("price")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="grade",
    y="price"
)
plt.title("price distribution")
plt.xlabel("grade")
plt.ylabel("price")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="waterfront",
    y="price"
)
plt.title("price distribution across waterfront")
plt.xlabel("waterfront")
plt.ylabel("price")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="long",
    y="lat",
    hue="price",
    palette="coolwarm",
    alpha=0.6,
    legend=False
)
plt.title("longitude vs latitude")
plt.xlabel("longitude")
plt.ylabel("latitude")
plt.show()
```

```python
location_summary = df.groupby("zipcode")["price"].agg(
    ["count", "mean", "median", "max"]
).reset_index()

location_summary = location_summary.sort_values(
    by="mean",
    ascending=False
)

location_summary.head(10)
```
